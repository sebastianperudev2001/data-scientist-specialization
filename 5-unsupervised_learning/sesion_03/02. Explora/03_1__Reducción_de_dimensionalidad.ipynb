### 1. Librerias

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import seaborn as sns
```

#### 1.1 Ejemplo:

```python
# 1. Crear datos simulados de clientes
np.random.seed(42)
dbClientes = pd.DataFrame({
    "Edad": np.random.randint(18, 65, 50),
    "Ingresos": np.random.randint(800, 5000, 50),
    "Gasto_online": np.random.randint(100, 3000, 50),
    "Visitas_web": np.random.randint(1, 30, 50)
})
```

```python
dbClientes.head()
```

```python
# 2. Estandarizar
X_scaled = StandardScaler().fit_transform(dbClientes)
X_scaled
```

```python
# 3. PCA con 2 componentes
pca = PCA(n_components=3)
# pca = PCA()
X_pca = pca.fit_transform(X_scaled)
```

```python
var_exp = pca.explained_variance_ratio_
var_cum = np.cumsum(var_exp)

plt.figure(figsize=(8,5))

# Barras
bars = plt.bar(range(1,len(var_exp)+1), var_exp*100, alpha=0.7, label="Varianza individual")

# Línea acumulada
plt.plot(range(1,len(var_exp)+1), var_cum*100, marker='o', color='red', label="Varianza acumulada")

# Etiquetas en las barras
for i, v in enumerate(var_exp*100):
    plt.text(i+1, v + 1, f"{v:.1f}%", ha='center', fontsize=9)

# Etiquetas en la curva acumulada
for i, v in enumerate(var_cum*100):
    plt.text(i+1, v + 1, f"{v:.1f}%", color='red', ha='center', fontsize=9)

# Ejes y título
plt.xlabel("Componente Principal")
plt.ylabel("% Varianza explicada")
plt.title("Scree Plot con barras y acumulado")
plt.legend()
plt.xticks(range(1,len(var_exp)+1))
plt.ylim(0, 110)   # para que no se corten los textos

plt.show()
```

```python
# 4. Coordenadas de las variables (cargas)
cargas = pca.components_.T * np.sqrt(pca.explained_variance_)
pd.DataFrame(cargas, index=dbClientes.columns, columns=["PC1","PC2","PC3"])
```

```python
# 5. Biplot
plt.figure(figsize=(8,6))

# Puntos (clientes)
sns.scatterplot(x=X_pca[:,0], y=X_pca[:,1], s=70, color="royalblue", edgecolor="k")

# Vectores (variables)
for i, var in enumerate(dbClientes.columns):
    plt.arrow(0, 0, cargas[i,0]*2, cargas[i,1]*2,
              color='red', alpha=0.8, head_width=0.05)
    plt.text(cargas[i,0]*2.2, cargas[i,1]*2.2, var, color='red', ha='center', va='center')

# Etiquetas y título
plt.xlabel("PC1 ({}%)".format(round(pca.explained_variance_ratio_[0]*100,2)))
plt.ylabel("PC2 ({}%)".format(round(pca.explained_variance_ratio_[1]*100,2)))
plt.title("Biplot de PCA - Clientes")
plt.axhline(0, color='grey', lw=1)
plt.axvline(0, color='grey', lw=1)
plt.grid(alpha=0.3)
plt.show()
```

```python
loadings = pd.DataFrame(cargas, index=dbClientes.columns, columns=["PC1","PC2","PC3"])
loadings.plot(kind="bar", figsize=(8,5))
plt.title("Contribución de variables a PC1, PC2 y PC3")
plt.show()
```
